```python
import torch
import math
import torch.nn.functional as F
from torch import Tensor
```

```python
# ce loss
def cross_entropy(
    input: Tensor, #[N,C]
    target: Tensor, #[N]
    ignore_index: int = -100,
) -> Tensor:
    valid_mask = target != ignore_index #[N]
    valid_input = input[valid_mask] #[N_valid, C]
    valid_target = target[valid_mask] #[N_valid]
    row_index = torch.arange(valid_input.size(0)) #[N_valid]
    log_prob = torch.log_softmax(valid_input,dim=1) #[N_valid, C]
    picked = log_prob[row_index,valid_target]
    res = -picked
    return res.mean()
```

```python
# mse loss
def mse_loss(
    input: Tensor, #[N,C]
    target: Tensor, #[N,C]
    ignore_index: int = -100,
) -> Tensor:
    valid_mask = target != ignore_index #[N,C]
    valid_input = input[valid_mask] #[N_valid]
    valid_target = target[valid_mask] #[N_valid]
    res = (valid_input - valid_target) ** 2
    return res.mean() 
```
